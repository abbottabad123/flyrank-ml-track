## 1. Question

*The research question and the decision it supports.*

### Research Question

FlyRank content teams need a practical way to prioritize existing pages for refresh or review. This study asks whether a machine-learning ranking workflow can identify content pages showing decline risk and place the most useful review candidates near the top of a prioritized queue.

### Decision Supported

The output is intended to help content teams decide which pages deserve human review first. It is a decision-support tool rather than an automatic publishing or search-ranking system.

## 2. Data

*Which release, which tables, date windows, what you excluded and why. Public-safe.*

### Dataset and Scope

This analysis uses the anonymized FlyRank content-refresh dataset included in the repository. The dataset contains 30,000 content records used to evaluate whether pages show signals associated with content decline.

The analysis uses public-safe, anonymized fields only. Client names, page titles, URLs, domains, private queries, and other identifying information were excluded from the presentation.

The data was prepared at the content-page level, with the relevant search and performance signals used to construct the modelling features and decline-risk label. Validation uses a client-holdout design so that evaluation reflects performance on clients not used for model fitting.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

### Method

The target is `is_declining_label`, representing whether a content page is identified as showing decline risk. The workflow compares a simple baseline ranking with candidate machine-learning models using the same client-holdout validation design.

The model uses available content and search-performance signals while excluding identifying fields. Leakage checks were used to avoid using information that would only be available after the prediction point.

Random Forest was selected as the strongest model using Precision@50. The model is used for ranking and prioritization, not for making causal claims about search-engine behavior or guaranteeing future traffic changes.

## 4. Results (vs baseline)

*Model vs baseline on the same split. The honest table.*

### Model vs Baseline

The Random Forest model performed substantially better than the baseline on the same client-holdout evaluation split.

| Model | Precision@50 | ROC AUC | Average Precision |
|---|---:|---:|---:|
| Baseline rules | 0.240 | — | — |
| Random Forest | **0.740** | **0.750** | **0.618** |

### Honest Result

Random Forest improved Precision@50 from 0.240 to 0.740, meaning the top 50 recommendations contained a much higher concentration of pages identified as showing decline risk.

However, these results measure predictive performance only. They do not prove that refreshing a page will improve its traffic or search performance, and the model should therefore be used as a prioritization aid for human review.

## 5. Limitations

*What this work cannot claim.*

### Limitations

This analysis shows predictive performance on the available anonymized dataset; it does not establish causal relationships.

The model does not prove that a content refresh will increase traffic, improve Google rankings, or change search-engine behavior. The results should also not be interpreted as a guarantee that every high-ranked page requires a refresh.

The ranked output is intended to support human decision-making. Content teams should review the page context, editorial quality, search intent, and other relevant information before taking action.

The dataset is anonymized and represents the available FlyRank release, so results may not generalize to every client, content type, or future time period.

## 6. Ranked recommendations

*The action playbook output — the paper's recommendations section.*

### Recommended Action Playbook

The ranked output should be used as a prioritized review queue rather than an automatic decision system.

1. **Start with the highest-ranked pages** — Review the pages with the strongest model scores first because they represent the model's highest-priority candidates.

2. **Verify the page manually** — Check the page's current content, search intent, relevance, and editorial quality before taking action.

3. **Choose the appropriate action** — Depending on the review, a page may need a content refresh, CTR review, engagement review, or no immediate action.

4. **Use confidence and reason signals** — Where available, use the model confidence and reason-code information to understand why an item was prioritized.

5. **Keep humans in the loop** — Do not automatically publish, delete, or substantially change content based only on the model score.

The main recommendation is to use the model as a decision-support tool that helps FlyRank teams focus their limited review time on the highest-priority content opportunities.
